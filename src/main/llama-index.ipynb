```python
import os
from google.colab import userdata
google_api_key = userdata.get("GOOGLE_API_KEY")
print(google_api_key)
os.environ["GOOGLE_API_KEY"] = google_api_key
import logging
import sys
logging.basicConfig(stream=sys.stdout, level=logging.DEBUG, force=True)

from llama_index.core import Settings
from llama_index.llms.gemini import Gemini
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

Settings.llm = Gemini(
    model_name = "models/gemini-flash-latest",
    safety_settings = {
        "HARM_CATEGORY_HARASSMENT": "BLOCK_NONE",
        "HARM_CATEGORY_HATE_SPEECH": "BLOCK_NONE",
        "HARM_CATEGORY_SEXUALLY_EXPLICIT": "BLOCK_NONE",
        "HARM_CATEGORY_DANGEROUS_CONTENT": "BLOCK_NONE"
    }
 )
Settings.embed_model = HuggingFaceEmbedding(
    model_name = "BAAI/bge-m3"
)
```

```python
from llama_index.core import SimpleDirectoryReader
documents=SimpleDirectoryReader(input_dir="/content/drive/MyDrive/github.com_gilbutITbook_080441/7_llama_index/data/").load_data()
```

```python
print(documents)

from llama_index.core import VectorStoreIndex
index = VectorStoreIndex.from_documents(documents)
```

```python
print(index)
query_index = index.
print(query_index.query("은비as_query_engine()의 나이는?"))
print(query_index.query("잭이라는 사람이 있는지? 있다면 그의 직업은 "))
print(query_index.query("How old is 은비?"))
print(query_index.query("은비와 용병이 쫒기는 이유를 말해줘"))
```

```python
index.storage_context.persist(persist_dir="./storage")
```

```python
response = Settings.llm.complete("한국에서 제일 높은산은 그리고 남한에서 제일 높은신도 알려주고 가장긴 강도 알려주세요")
print(str(response))
embeddings=Settings.embed_model.get_text_embedding(str(response))
print(len(embeddings))
print(embeddings[:5])
```
